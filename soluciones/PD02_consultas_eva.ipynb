## 1. Preparación de columnas · 10 minutos

Inspeccionar la tabla y crear `codigo_departamento` a partir de `Código Dane departamento`, `anio` desde `Año`, `area_ha` desde `Área cosechada` y `produccion_t` desde `Producción`. El código se conserva como texto; año, área y producción se convierten a tipos numéricos. Registrar las ausencias resultantes.

📘 **Apoyo:** `.str.strip`, `pd.to_numeric`, `astype("Int64")` e `isna`.

✅ **Comprobación:** Las columnas originales se conservan y no se eliminan filas. El corte de entrada contiene 166.732 registros.

````python
# Desarrollo de la actividad 1
# Preparar las cuatro columnas de trabajo en eva.
eva['codigo_departamento'] = eva['Código Dane departamento'].astype(str).str.strip()
anio_limpio = eva['Año'].astype(str).str.strip().str.replace(',', '.', regex=False)
eva['anio'] = pd.to_numeric(anio_limpio, errors='coerce').astype('Int64')
area_limpio = eva['Área cosechada'].astype(str).str.strip().str.replace(',', '.', regex=False)
eva['area_ha'] = pd.to_numeric(area_limpio, errors='coerce')
prod_limpio = eva['Producción'].astype(str).str.strip().str.replace(',', '.', regex=False)
eva['produccion_t'] = pd.to_numeric(prod_limpio, errors='coerce')
print("Tipos de datos resultantes:")
display(eva[['codigo_departamento', 'anio', 'area_ha', 'produccion_t']].dtypes)
print("\n--- Diagnóstico de valores que no pudieron convertirse ---")
orig_anio_vacio = eva['Año'].isna() | (eva['Año'].astype(str).str.strip() == '')
fallos_anio = eva['anio'].isna() & ~orig_anio_vacio
print(f"Total de conversiones fallidas en 'anio': {fallos_anio.sum()}")
if fallos_anio.sum() > 0:
    display(eva.loc[fallos_anio, ['Año', 'anio']].head(5))
orig_area_vacio = eva['Área cosechada'].isna() | (eva['Área cosechada'].astype(str).str.strip() == '')
fallos_area = eva['area_ha'].isna() & ~orig_area_vacio
print(f"Total de conversiones fallidas en 'area_ha': {fallos_area.sum()}")
if fallos_area.sum() > 0:
    display(eva.loc[fallos_area, ['Área cosechada', 'area_ha']].head(5))
orig_prod_vacio = eva['Producción'].isna() | (eva['Producción'].astype(str).str.strip() == '')
fallos_prod = eva['produccion_t'].isna() & ~orig_prod_vacio
print(f"Total de conversiones fallidas en 'produccion_t': {fallos_prod.sum()}")
if fallos_prod.sum() > 0:
    display(eva.loc[fallos_prod, ['Producción', 'produccion_t']].head(5))

# Mostrar los tipos y los valores que no pudieron convertirse.
print(f"\n[Comprobación] Total de filas conservadas: {eva.shape[0]} (Debe ser 166.732)")
````

**Interpretación del resultado:**

Se aila el codigo del departamento como texto puro, en cambio año, area cosechada y produccion total se transforman a numéricos, se permite identificar y aislar los registros con caracteres alfanumericos o anomalías que no se hayan podido convertir realizando la convercion a nulos sin alterar o elimnar las filas del dataset.

## 2. Selección con máscara y loc · 15 minutos

Seleccionar el departamento `15` (Boyacá), los años 2023 y 2024, área cosechada positiva y producción no negativa. Guardar todas las columnas de las filas seleccionadas en `consulta_loc`, utilizando `.copy()`. Presentar el total original, el seleccionado y la diferencia.

📘 **Apoyo:** `.eq`, `.isin`, `.gt`, `.ge`, `&` y `.loc`.

✅ **Comprobación:** Todas las filas seleccionadas cumplen simultáneamente las cuatro condiciones. La diferencia respecto a la entrada incluye registros fuera del alcance territorial o temporal; no implica que todos sean errores.

````python
# Desarrollo de la actividad 2
# Construir la máscara y guardar consulta_loc.
cond_depto = eva['codigo_departamento'].eq('15')
cond_anios = eva['anio'].isin([2023, 2024])
cond_area = eva['area_ha'].gt(0)
cond_prod = eva['produccion_t'].ge(0)
mascara = cond_depto & cond_anios & cond_area & cond_prod
consulta_loc = eva.loc[mascara].copy()
total_original = len(eva)
total_seleccionado = len(consulta_loc)
diferencia = total_original - total_seleccionado
print(f"Total de filas original: {total_original}")
print(f"Total de filas seleccionadas (consulta_loc): {total_seleccionado}")
print(f"Diferencia (registros excluidos): {diferencia}")
# Mostrar los conteos de entrada, selección y diferencia
display(consulta_loc.head())
````

**Interpretación del resultado:**

se realiza una mascara booleana combinando cuatro condiciones mediante el operador logico &, se extraajeron todas las columnas de las filas en base a que cumplan de manera simultanea los cuales se almacenan en un nuevo dataframe lo cual garantiza la integridad.
Se evidencia que la diferencia entre los registros total y seleccionado refleja la cantidad de observaciones excluidas ya que estas pertenecen a otros departamentos, diferentes periodos de tiempo o no cumplen con los rangos porductivos especificados.

## 3. Consulta equivalente con query · 10 minutos

Repetir exactamente la selección anterior mediante `.query`, con `engine="python"`, y guardar el resultado en `consulta_query`. Usar una variable para el código departamental y referenciarla con `@`. Comparar ambos DataFrames sin eliminar columnas.

📘 **Apoyo:** `.query`, `@variable` y `pd.testing.assert_frame_equal`.

✅ **Comprobación:** La comparación debe confirmar las mismas filas, columnas y valores. Si falla, se revisan las condiciones y los tipos antes de continuar.

````python
# Desarrollo de la actividad 3
# Construir consulta_query.
import pandas as pd
depto_code = '15'
consulta_query = eva.query(
    "codigo_departamento == @depto_code and anio in [2023, 2024] and area_ha > 0 and produccion_t >= 0",
    engine="python"
).copy()
# Comprobar que consulta_loc y consulta_query son equivalentes.
pd.testing.assert_frame_equal(consulta_loc, consulta_query)
print("[Comprobación Exitosa] consulta_loc y consulta_query son idénticos en filas, columnas y valores.")
````

**Interpretación del resultado:**

Se procede a replicar el subconjuto de datos de manera limpia y legible, se usan las expresiones de consulta de cadena con python, la prueba afirmo sin errores que el filtrado y la máscara arrojan los mismos resultados sin realizar cambios a las columnas y al contenido de registros.

## 4. Conteos y porcentajes · 20 minutos

Sobre `consulta_loc`, construir `resumen` con una fila por año y cultivo y una columna `registros`. Añadir el total de registros de cada año y `porcentaje_anual`. Ordenar los resultados por año y cantidad de registros descendente. Conservar los grupos con claves ausentes mediante `dropna=False`.

📘 **Apoyo:** `groupby`, `size`, `reset_index`, `transform("sum")` y `sort_values`.

✅ **Comprobación:** La suma de registros coincide con el tamaño de consulta_loc. Los porcentajes suman aproximadamente 100 % por año; se admite una pequeña tolerancia numérica y se redondea solo para mostrar.

````python
# Desarrollo de la actividad 4
# Crear resumen con registros, total_anio y porcentaje_anual.
resumen = consulta_loc.groupby(['anio', 'Cultivo'], dropna=False).size().reset_index(name='registros')
resumen['total_anio'] = resumen.groupby('anio')['registros'].transform('sum')
resumen['porcentaje_anual'] = (resumen['registros'] / resumen['total_anio']) * 100
resumen = resumen.sort_values(by=['anio', 'registros'], ascending=[True, False])
print("Primeros registros del resumen por año y cultivo:")
display(resumen.head(10))
total_registros_resumen = resumen['registros'].sum()
assert total_registros_resumen == len(consulta_loc), "¡Error! La suma de registros del resumen no coincide con el tamaño de consulta_loc."

# Verificar los totales y los porcentajes por año.
print(f"\n[Comprobación Exitosa] La suma de registros ({total_registros_resumen}) coincide exactamente con el tamaño de consulta_loc.")
````

**Interpretación del resultado:**

se consolida el conjuto para garantizar que las categorias o valores ausentes fueran descartados en el analisis, se realiza la transformacion y se incorpora el volumen global de registros por periodo el cual facilial el calculo del porcentaje, se ordenan los datos jerarquicamente por año de manera descendente por la cantidad de registros para validar el total de observaciones sin perdidas numericas.

## 5. Tabla dinámica y conclusiones · 15 minutos

Construir una tabla dinámica a partir de `resumen`, con cultivos en filas, años en columnas y conteos como valores. Completar con cero las combinaciones sin registros. Exportar `resumen` a `OUT + "/resumen_eva.csv"`, sin índice, y redactar dos conclusiones.

📘 **Apoyo:** `pivot_table`, `fill_value=0` y `to_csv`.

✅ **Comprobación:** Los totales de la tabla dinámica concilian con resumen. Las conclusiones se refieren a la cantidad de registros y no a la participación de cada cultivo en la superficie agrícola.

````python
# Desarrollo de la actividad 5
# Construir y mostrar la tabla dinámica
import os
tabla_dinamica = resumen.pivot_table(
    index='Cultivo',
    columns='anio',
    values='registros',
    fill_value=0
)

print("Tabla dinámica de registros por cultivo y año:")
display(tabla_dinamica)
os.makedirs(OUT, exist_ok=True)
ruta_resumen = os.path.join(OUT, "resumen_eva.csv")
resumen.to_csv(ruta_resumen, index=False)

# Exportar resumen y explicar dos resultados en la celda siguiente.
print(f"\n[Comprobación Exitosa] Archivo 'resumen_eva.csv' exportado correctamente en: {ruta_resumen}")
````

**Interpretación del resultado:**

La tabla permite identificar que el grupo reducido de cultivos concentra la mayor cantidad de registros reportados en el departamento de Boyacá para el periodo 2023-2024 lo cual refleja heterogeneidad en el reporte de productividad.
Al comparar los registros para los años 2023 y 2024 por medio de la estructura matricial se logra observar la estabilidad o variacion en las entradas por cultivo, la estructuracion facilito la deteccion de periodos con ausencias totales de registros.
