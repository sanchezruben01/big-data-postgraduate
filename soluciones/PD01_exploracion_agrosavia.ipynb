## 2. Diagnóstico de cinco campos · 10 minutos

Construir una tabla llamada `perfil` para `Secuencial`, `Fecha de Análisis`, `Departamento`, `Municipio` y `pH agua:suelo`. Para cada campo se registran los nulos de pandas, los textos vacíos después de retirar espacios y los marcadores `ND`. Las tres situaciones se cuentan por separado.

📘 **Apoyo:** `isna`, `.str.strip`, `.str.upper`, `.eq` y `sum`.

✅ **Comprobación:** `perfil` debe tener una fila por campo. Con `keep_default_na=False`, un texto vacío o `ND` no se convierte automáticamente en un nulo.

````python
# Desarrollo de la actividad 2
# Crear y mostrar perfil con las cinco columnas solicitadas.
import pandas as pd
campos = ['Secuencial', 'Fecha de Análisis', 'Departamento', 'Municipio', 'pH agua:suelo']
filas_perfil = []
for col in campos:
    nulos_pandas = suelo[col].isna().sum()
    textos_vacios = (suelo[col].astype(str).str.strip() == '').sum()
    marcadores_nd = (suelo[col].astype(str).str.strip().str.upper() == 'ND').sum() 
    filas_perfil.append({
        'Campo': col,
        'Nulos_Pandas': nulos_pandas,
        'Textos_Vacios': textos_vacios,
        'ND': marcadores_nd
    })
perfil = pd.DataFrame(filas_perfil)
display(perfil)
````

**Interpretación del resultado:**

La tabla contirnr una fila por cada uno de los 5 campos analizados lo cual permite verificar de forma independiente cada anomalia, debido a que se conservan los tipos de texto originales se detecta con precision los espacios vacios y los textos ND lo cual no habria sido posible al convertir el texto.

## 3. Conversión de pH y fecha · 15 minutos

Crear `ph` desde `pH agua:suelo` y `fecha_dt` desde `Fecha de Análisis`. Utilizar conversión numérica con `errors="coerce"` y fecha con formato `%d/%m/%Y`. Conservar los originales. Distinguir entradas vacías de textos no vacíos que no pudieron convertirse.

📘 **Apoyo:** `pd.to_numeric`, `.str.replace`, `pd.to_datetime`, `isna` y máscaras booleanas.

✅ **Comprobación:** Se muestran los conteos de conversiones fallidas y hasta cinco ejemplos con su valor original. No se reemplazan medidas ausentes por cero.

````python
# Desarrollo de la actividad 3
# Crear suelo["ph"] y suelo["fecha_dt"].
ph_limpio = suelo['pH agua:suelo'].astype(str).str.replace(',', '.', regex=False)
suelo['ph'] = pd.to_numeric(ph_limpio, errors='coerce')
suelo['fecha_dt'] = pd.to_datetime(suelo['Fecha de Análisis'], format='%d/%m/%Y', errors='coerce')

# Mostrar los fallos de conversión y sus valores originales.
es_nulo_orig_ph = suelo['pH agua:suelo'].isna() | (suelo['pH agua:suelo'].astype(str).str.strip() == '')
fallo_ph = suelo['ph'].isna() & ~es_nulo_orig_ph
print(f"Total de conversiones fallidas en 'ph': {fallo_ph.sum()}")
print("Hasta 5 ejemplos originales con fallo en 'ph':")
display(suelo.loc[fallo_ph, ['pH agua:suelo', 'ph']].head(5))
es_nulo_orig_fecha = suelo['Fecha de Análisis'].isna() | (suelo['Fecha de Análisis'].astype(str).str.strip() == '')
fallo_fecha = suelo['fecha_dt'].isna() & ~es_nulo_orig_fecha
print(f"\nTotal de conversiones fallidas en 'fecha_dt': {fallo_fecha.sum()}")
print("Hasta 5 ejemplos originales con fallo en 'fecha_dt':")
display(suelo.loc[fallo_fecha, ['Fecha de Análisis', 'fecha_dt']].head(5))
````

**Interpretación del resultado:**

Se realiza la conversion de los valores incompatibles en nulos para no interrumpir la ejecucion, tambien se usan mascaras booleanas para separar los campos que venian vacios o nulos, se mantienen los datos originales y se reemplazan los campos ausentes pos ceros manteniendo la cantidad de datos faltantes.

## 4. Registros para revisión · 15 minutos

Crear la columna booleana `en_revision`. Se marca una fila si su pH es ausente, está fuera de `[0, 14]` o su fecha no se pudo interpretar. Contar por separado cada motivo y el total de filas marcadas. Conservar todas las observaciones.

📘 **Apoyo:** `between`, `isna`, `|`, `~`, `.loc` y `value_counts`.

✅ **Comprobación:** Las filas marcadas y no marcadas deben sumar el total original. Una misma fila puede tener varios motivos; por eso la suma de los motivos no necesariamente coincide con el total marcado.

````python
# Desarrollo de la actividad 4
# Crear en_revision y los conteos por motivo.
ph_ausente = suelo['ph'].isna()
ph_fuera_rango = ~suelo['ph'].between(0, 14) & ~ph_ausente 
fecha_fallida = suelo['fecha_dt'].isna() & ~(suelo['Fecha de Análisis'].isna() | (suelo['Fecha de Análisis'].astype(str).str.strip() == ''))
conteo_ph_ausente = ph_ausente.sum()
conteo_ph_fuera_rango = ph_fuera_rango.sum()
conteo_fecha_fallida = fecha_fallida.sum()
print("Conteo por motivo:")
print(f"- pH ausente: {conteo_ph_ausente}")
print(f"- pH fuera de rango [0, 14]: {conteo_ph_fuera_rango}")
print(f"- Fecha no interpretable: {conteo_fecha_fallida}")
suelo['en_revision'] = ph_ausente | ph_fuera_rango | fecha_fallida
total_marcadas = suelo['en_revision'].sum()
total_no_marcadas = (~suelo['en_revision']).sum()
print(f"\nTotal de filas marcadas para revisión: {total_marcadas}")
print(f"Total de filas no marcadas: {total_no_marcadas}")
# Comprobar la conservación del total original con assert.
assert (total_marcadas + total_no_marcadas) == total_filas_original, "¡Error! Las filas marcadas y no marcadas no suman el total original."
print("\n[Comprobación Exitosa] Las filas marcadas y no marcadas suman exactamente el total original de filas.")

````

**Interpretación del resultado:**

se crea una variable booleana para identificar las filas que representan anomalias críticas en las mediciones. se realiza el conteo de las incidencias por separado dado a que las filas puedan incurrir con fallos simultaneos, esto en base a conservar la integridad de las observaciones originales y garantizar que no se eliminen las filas.

## 5. Evidencias y conclusiones · 10 minutos

Construir `controles` con las columnas `control` y `cantidad`: total de entrada, pH sin número, pH fuera del intervalo, fecha no interpretable y total de filas en revisión. Exportar a `OUT + "/controles_agrosavia.csv"` sin índice. Redactar dos hallazgos apoyados en los conteos.

📘 **Apoyo:** `pd.DataFrame`, `to_csv(index=False)` y las tablas de las actividades anteriores.

✅ **Comprobación:** El archivo se genera dentro de la ruta `OUT`. Cada hallazgo menciona un conteo y explica su significado o limitación.

````python
# Desarrollo de la actividad 5
# Construir controles, mostrarlo y exportarlo a OUT.
import os
datos_controles = [
    {"control": "Total de entrada", "cantidad": total_filas_original},
    {"control": "pH sin número", "cantidad": fallo_ph.sum()},
    {"control": "pH fuera del intervalo", "cantidad": conteo_ph_fuera_rango},
    {"control": "Fecha no interpretable", "cantidad": conteo_fecha_fallida},
    {"control": "Total de filas en revisión", "cantidad": total_marcadas}
]
controles = pd.DataFrame(datos_controles)
print("Tabla de controles generada:")
display(controles)
os.makedirs(OUT, exist_ok=True)
ruta_salida = os.path.join(OUT, "controles_agrosavia.csv")
controles.to_csv(ruta_salida, index=False)
print(f"\n[Comprobación Exitosa] Archivo exportado correctamente en: {ruta_salida}")
````

**Interpretación del resultado:**

se consolida la tabla controles el cual recopila las metricas globales del proceso de auditoria, realiza la exportacion, realiza el conteo de los registros, se evidencia la presencia de errores tipograficos o mediciones imposibles en campo.
